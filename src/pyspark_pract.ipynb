```python
data = [
    (101, "John", ["Laptop","Mouse","Keyboard"]),
    (102, "Alice", ["Phone","Charger","Phone"]),
    (103, "Bob", ["Headphone"]),
    (104, "David", []),
    (105, "Emma", ["Laptop","Laptop","Mouse"]),
    (106, "Tom", ["Tablet","Pen","Cover"])
]
df=spark.createDataFrame(data,["id","customer","products"])
df.show(truncate=False)
```

```python
df.filter(array_contains("products","Laptop")).select("customer",size("products").alias("total_products")).show()
```

```python
'''r1=df.select("customer",array_distinct("products").alias("products"),size(""))'''
r=df.withColumn("nproducts",array_distinct("products"))
r.select("customer","nproducts",size("nproducts").alias("unique_count")).show(truncate=False)
```

```python

df.withColumn("updated_products",array_remove("products","Mouse")).withColumn("remaining_products",size("updated_products")).select("customer","updated_products","remaining_products").show(truncate=False)
```
